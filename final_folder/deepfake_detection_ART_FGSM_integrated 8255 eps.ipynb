```python
import os
import glob
import random
import cv2
import numpy as np
from PIL import Image
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
import torchvision.transforms as transforms
import timm
```

```python
# 1. Device setup
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")
```

```python
# 2. Aggressive Social Media Compression Function
def apply_social_media_compression(img_bgr):
    """
    Simulates aggressive social media pipeline: 
    Downscale -> Chroma Subsampling (JPEG) -> Gaussian Blur -> Upscale
    """
    h, w, _ = img_bgr.shape
    
    # Step 1: Aggressive Downscaling (mimicking platform resizing)
    scale_factor = random.uniform(1.0, 1.0)
    new_h, new_w = int(h * scale_factor), int(w * scale_factor)
    img_resized = cv2.resize(img_bgr, (new_w, new_h), interpolation=cv2.INTER_AREA)
    img_resized = cv2.resize(img_resized, (w, h), interpolation=cv2.INTER_LINEAR)
    
    # Step 2: Apply a mild Gaussian Blur
    if random.random() < 0.7:
        img_resized = cv2.GaussianBlur(img_resized, (5, 5), 0)
        
    # Step 3: Extreme JPEG Compression with Chroma Subsampling
    quality = random.randint(10, 25)
    encode_param = [int(cv2.IMWRITE_JPEG_QUALITY), quality]
    _, encimg = cv2.imencode('.jpg', img_resized, encode_param)
    decimg_bgr = cv2.imdecode(encimg, 1)
    
    # Convert to RGB for PIL / PyTorch
    return cv2.cvtColor(decimg_bgr, cv2.COLOR_BGR2RGB)
```

```python
# 3. Balanced Dataset Class with Compression Support
class BalancedDeepfakeDataset(Dataset):
    def __init__(self, split='train', base_dir='/kaggle/input', transform=None, seed=42, compression_prob=0.5):
        self.transform = transform
        self.compression_prob = compression_prob
        self.image_paths = []
        self.labels = []
        self.rng = random.Random(seed)
        
        self._temp_data = {
            'celeb': {0: [], 1: []},
            '140k': {0: [], 1: []},
            'ffpp': {0: [], 1: []}
        }
        
        print(f"[{split.upper()}] Scanning {base_dir} to find datasets...")
        dataset_paths = self._auto_discover_paths(base_dir)
        
        self._load_celebdf(dataset_paths['celeb'], split)
        self._load_140k(dataset_paths['140k'], split)
        self._load_ffpp(dataset_paths['ffpp'], split)
        
        self._balance_and_merge(split)

    def _auto_discover_paths(self, base_dir):
        paths = {'celeb': None, 'ffpp': None, '140k': None}
        for root, dirs, files in os.walk(base_dir):
            if 'Celeb_V2' in dirs and not paths['celeb']:
                paths['celeb'] = os.path.join(root, 'Celeb_V2')
            if 'dataset_processed_split' in dirs and not paths['ffpp']:
                paths['ffpp'] = os.path.join(root, 'dataset_processed_split')
            if 'real-vs-fake' in dirs and not paths['140k']:
                paths['140k'] = os.path.join(root, 'real-vs-fake')
        return paths

    def _add_from_folder(self, dataset_name, folder_path, label):
        if os.path.exists(folder_path):
            for ext in ('*.png', '*.jpg', '*.jpeg'):
                paths = glob.glob(os.path.join(folder_path, ext))
                self._temp_data[dataset_name][label].extend(paths)

    def _load_celebdf(self, base_path, split):
        split_map = {'train': 'Train', 'val': 'Val', 'test': 'Test'}
        target_split = split_map.get(split)
        if base_path and os.path.exists(base_path) and target_split:
            self._add_from_folder('celeb', os.path.join(base_path, target_split, 'real'), 0)
            self._add_from_folder('celeb', os.path.join(base_path, target_split, 'fake'), 1)

    def _load_140k(self, base_path, split):
        split_map = {'train': 'train', 'val': 'valid', 'test': 'test'}
        target_split = split_map.get(split)
        if base_path and os.path.exists(base_path) and target_split:
            self._add_from_folder('140k', os.path.join(base_path, target_split, 'real'), 0)
            self._add_from_folder('140k', os.path.join(base_path, target_split, 'fake'), 1)

    def _load_ffpp(self, base_path, split):
        if base_path:
            split_path = os.path.join(base_path, split)
            if os.path.exists(split_path):
                for class_dir in os.listdir(split_path):
                    dir_path = os.path.join(split_path, class_dir)
                    if os.path.isdir(dir_path):
                        label = 0 if class_dir == 'Real' else 1
                        self._add_from_folder('ffpp', dir_path, label)

    def _balance_and_merge(self, split):
        min_count = float('inf')
        for ds_name, labels in self._temp_data.items():
            count_real = len(labels[0])
            count_fake = len(labels[1])
            if count_real < min_count: min_count = count_real
            if count_fake < min_count: min_count = count_fake
                    
        if min_count == 0:
            print(f"❌ ERROR: One or more categories in {split.upper()} has 0 images. Cannot balance.")
            return

        print(f"[{split.upper()}] Balancing to {min_count} images per category...")
        
        for ds_name, labels in self._temp_data.items():
            for label, paths in labels.items():
                sampled_paths = self.rng.sample(paths, min_count)
                self.image_paths.extend(sampled_paths)
                self.labels.extend([label] * min_count)
        
        combined = list(zip(self.image_paths, self.labels))
        if not combined: return

        self.rng.shuffle(combined)
        self.image_paths, self.labels = zip(*combined)
        self.image_paths = list(self.image_paths)
        self.labels = list(self.labels)
        
        print(f"✅ Final {split} dataset size: {len(self.image_paths)} images ready.\n")

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):
        img_path = self.image_paths[idx]
        label = self.labels[idx]
        
        # Load image via CV2 to match the compression pipeline
        img_bgr = cv2.imread(img_path)
        
        # Dynamically apply our compression function
        if self.rng.random() < self.compression_prob:
            img_rgb = apply_social_media_compression(img_bgr)
        else:
            img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
            
        # Convert to PIL for Torchvision transforms
        img_pil = Image.fromarray(img_rgb)
        
        if self.transform:
            img_tensor = self.transform(img_pil)
            
        return img_tensor, torch.tensor(label, dtype=torch.long)
```

```python
# 4. Define Transforms for Xception (299x299)
def get_data_transforms(img_size=299):
    norm_mean = [0.485, 0.456, 0.406]
    norm_std = [0.229, 0.224, 0.225]

    train_transform = transforms.Compose([
        transforms.Resize((img_size, img_size)),
        transforms.RandomCrop(img_size, padding=4, padding_mode='reflect'),
        transforms.RandomHorizontalFlip(p=0.5),
        transforms.RandomRotation(degrees=15),
        transforms.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.1, hue=0.05),
        transforms.ToTensor(),
        transforms.Normalize(mean=norm_mean, std=norm_std)
    ])

    val_transform = transforms.Compose([
        transforms.Resize((img_size, img_size)),
        transforms.ToTensor(),
        transforms.Normalize(mean=norm_mean, std=norm_std)
    ])

    return train_transform, val_transform
```

```python
# 5. Initialize Datasets and DataLoaders
train_transform, val_transform = get_data_transforms(img_size=299)

# Set compression probability to 0.7 for training, 0.0 for validation/testing
train_dataset = BalancedDeepfakeDataset(split='train', transform=train_transform, seed=42, compression_prob=0.7)
val_dataset   = BalancedDeepfakeDataset(split='val', transform=val_transform, seed=42, compression_prob=0.0)
test_dataset  = BalancedDeepfakeDataset(split='test', transform=val_transform, seed=42, compression_prob=0.0)

train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True, num_workers=2)
val_loader   = DataLoader(val_dataset, batch_size=32, shuffle=False, num_workers=2)
test_loader  = DataLoader(test_dataset, batch_size=32, shuffle=False, num_workers=2)
```

```python
class SpatialDeepfakeDetector(nn.Module):
    def __init__(self, pretrained=True):
        super(SpatialDeepfakeDetector, self).__init__()
        
        # Load Xception
        self.backbone = timm.create_model('xception', pretrained=pretrained)
        
        # Replace the classifier head for Binary Classification (1 output node)
        num_ftrs = self.backbone.fc.in_features
        self.backbone.fc = nn.Sequential(
            nn.Linear(num_ftrs, 512),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(512, 1) 
        )

    def forward(self, x):
        return self.backbone(x)
```

```python
# Initialize the model and move it to the GPU
model = SpatialDeepfakeDetector(pretrained=True)
model = model.to(device)

# Test the model output shape
images = images.to(device) 
with torch.no_grad():
    outputs = model(images)

print(f"Output tensor shape: {outputs.shape}")
print("If you see torch.Size([32, 1]), the Xception model is ready!")
```

```python
import torch.optim as optim
from tqdm import tqdm
import copy
from torch.optim.lr_scheduler import ReduceLROnPlateau
```

```python
# 1. Define Loss Function
criterion = nn.BCEWithLogitsLoss()
```

```python
# 2. Define Optimizer (Split learning rates)
backbone_params = []
classifier_params = []
for name, param in model.named_parameters():
    if "backbone.fc" in name:
        classifier_params.append(param)
    else:
        backbone_params.append(param)

optimizer = optim.AdamW([
    {'params':  backbone_params, 'lr': 1e-5}, # Slow updates to features
    {'params': classifier_params, 'lr': 1e-3}       # Fast updates to the head
])
```

```python
# 3. Early Stopping Class
class EarlyStopping:
    def __init__(self, patience=5, min_delta=0, verbose=True):
        self.patience = patience
        self.min_delta = min_delta
        self.verbose = verbose
        self.counter = 0
        self.best_loss = float('inf')
        self.early_stop = False
        self.best_model_wts = None

    def __call__(self, val_loss, model):
        if val_loss < self.best_loss - self.min_delta:
            if self.verbose:
                print(f"Validation loss decreased ({self.best_loss:.4f} --> {val_loss:.4f}). Saving model weights...")
            self.best_loss = val_loss
            self.best_model_wts = copy.deepcopy(model.state_dict())
            self.counter = 0
        else:
            self.counter += 1
            print(f"EarlyStopping counter: {self.counter} out of {self.patience}")
            if self.counter >= self.patience:
                self.early_stop = True
```

## Phase 2 addition: ART-based FGSM adversarial training

This section integrates **Adversarial Robustness Toolbox (ART)** into the existing Xception detector. The model architecture, dataset discovery/balancing, transforms, optimizer, loss, and original test evaluation are retained. FGSM is generated from the current model during training; a configurable subset of each batch is used for adversarial examples to help keep Kaggle runs practical. The final evaluation reports both clean test accuracy and FGSM-attacked test accuracy.

Runtime depends on the Kaggle GPU, dataset size, and dataloader speed; a hard wall-clock limit cannot be guaranteed. The training loop below uses a time budget and early stopping to avoid intentionally running beyond a 12-hour session.

```python
# ART integration (added for Phase 2)
# If ART is not installed in the Kaggle image, install it once. This requires internet access.
try:
    from art.estimators.classification import PyTorchClassifier
    from art.attacks.evasion import FastGradientMethod
except ImportError:
    %pip -q install adversarial-robustness-toolbox
    from art.estimators.classification import PyTorchClassifier
    from art.attacks.evasion import FastGradientMethod

import time, gc

# Model returns one logit (B, 1) for BCEWithLogitsLoss. ART's classifier adapter
# exposes equivalent two-class logits while preserving gradients to the same model.
class ARTBinaryLogitAdapter(nn.Module):
    def __init__(self, binary_model):
        super().__init__()
        self.binary_model = binary_model
    def forward(self, x):
        z = self.binary_model(x).reshape(-1, 1)
        return torch.cat((-0.5 * z, 0.5 * z), dim=1)

art_adapter = ARTBinaryLogitAdapter(model).to(device)
# Inputs are ImageNet-normalized. eps=0.01 is approximately a 2/255 pixel-space
# perturbation for typical ImageNet standard deviations; clip range is conservative.
ART_EPS = 0.01
ART_FGSM = FastGradientMethod(
    estimator=PyTorchClassifier(
        model=art_adapter,
        loss=nn.CrossEntropyLoss(),
        optimizer=optimizer,
        input_shape=(3, 299, 299),
        nb_classes=2,
        clip_values=(-3.0, 3.0),
        device_type='gpu' if device.type == 'cuda' else 'cpu',
    ),
    eps=ART_EPS,
    eps_step=ART_EPS,
    batch_size=16,
)

# Time guard: leave a buffer for final evaluation/checkpoint writing.
KAGGLE_SESSION_BUDGET_HOURS = 11.0
TRAIN_START_TIME = time.time()
TRAIN_TIME_BUDGET_SECONDS = KAGGLE_SESSION_BUDGET_HOURS * 3600
ART_ADV_FRACTION = 0.5       # generate FGSM for half of each training batch
ART_ADV_LOSS_WEIGHT = 0.5    # clean loss remains the main objective
print(f'ART FGSM ready | eps={ART_EPS} in normalized tensor space | adversarial fraction={ART_ADV_FRACTION:.0%}')
```

```python
# 4. Training Loop Setup — original clean training plus ART FGSM adversarial examples
# The existing model, loss, optimizer, scheduler and EarlyStopping class are unchanged.
epochs = 10
early_stopper = EarlyStopping(patience=3, verbose=True)
scheduler = ReduceLROnPlateau(optimizer, mode='min', patience=2, factor=0.1)

for epoch in range(epochs):
    if time.time() - TRAIN_START_TIME >= TRAIN_TIME_BUDGET_SECONDS:
        print('12-hour Kaggle safety guard reached its training budget; stopping before final evaluation.')
        break
    model.train()
    running_loss = 0.0
    clean_loss_total = 0.0
    adv_batches = 0
    loop = tqdm(train_loader, total=len(train_loader), leave=True, miniters=100)

    for batch_idx, (inputs, labels) in enumerate(loop):
        if time.time() - TRAIN_START_TIME >= TRAIN_TIME_BUDGET_SECONDS:
            print('Training time budget reached mid-epoch; ending epoch early.')
            break
        inputs = inputs.to(device)
        labels = labels.to(device).view(-1, 1).float()
        optimizer.zero_grad(set_to_none=True)

        # Preserve the original clean objective.
        clean_outputs = model(inputs)
        clean_loss = criterion(clean_outputs, labels)
        loss = clean_loss

        # ART attack generation uses integer class labels and the adapter over the same model.
        # Generate on a subset to reduce the additional compute cost.
        n_adv = max(1, int(inputs.size(0) * ART_ADV_FRACTION))
        x_adv_source = inputs[:n_adv].detach()
        y_adv_int = labels[:n_adv].view(-1).long()
        try:
            adv_np = ART_FGSM.generate(
                x=x_adv_source.detach().cpu().numpy().astype(np.float32),
                y=y_adv_int.detach().cpu().numpy().astype(np.int64)
            )
            adv_inputs = torch.from_numpy(adv_np).to(device=device, dtype=inputs.dtype)
            adv_outputs = model(adv_inputs)
            adv_loss = criterion(adv_outputs, labels[:n_adv])
            loss = (1.0 - ART_ADV_LOSS_WEIGHT) * clean_loss + ART_ADV_LOSS_WEIGHT * adv_loss
            adv_batches += 1
        except RuntimeError as exc:
            # Fail safely on occasional GPU memory pressure: complete the clean step.
            if 'out of memory' in str(exc).lower():
                if device.type == 'cuda': torch.cuda.empty_cache()
                print('ART batch skipped due to memory pressure; using clean loss for this batch.')
                loss = clean_loss
            else:
                raise

        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=5.0)
        optimizer.step()
        running_loss += float(loss.detach().item())
        clean_loss_total += float(clean_loss.detach().item())
        loop.set_description(f'Epoch [{epoch+1}/{epochs}] ART-FGSM')

    completed_batches = max(1, batch_idx + 1)
    avg_train_loss = running_loss / completed_batches
    print(f'\nEpoch {epoch+1}/{epochs} | combined loss={avg_train_loss:.4f} | clean loss={clean_loss_total/completed_batches:.4f} | ART batches={adv_batches}')

    # Validation is clean and unchanged; use it for checkpoint selection / early stopping.
    model.eval()
    val_loss, correct, total = 0.0, 0, 0
    with torch.no_grad():
        for inputs, labels in tqdm(val_loader, desc='Validation', leave=False, miniters=100):
            inputs = inputs.to(device)
            labels = labels.to(device).view(-1, 1).float()
            outputs = model(inputs)
            val_loss += criterion(outputs, labels).item()
            predictions = (torch.sigmoid(outputs) > 0.5).float()
            correct += (predictions == labels).sum().item()
            total += labels.size(0)
    avg_val_loss = val_loss / max(1, len(val_loader))
    val_accuracy = 100.0 * correct / max(1, total)
    print(f'Epoch {epoch+1}/{epochs} Validation - Loss: {avg_val_loss:.4f} | Acc: {val_accuracy:.2f}%')
    torch.save(model.state_dict(), f'xception_deepfake_detector_art_epoch_{epoch+1}.pth')
    scheduler.step(avg_val_loss)
    early_stopper(avg_val_loss, model)
    if early_stopper.early_stop:
        print(f'Early stopping triggered at epoch {epoch+1}.')
        break

if early_stopper.best_model_wts is not None:
    print('Loading best weights selected by clean validation loss...')
    model.load_state_dict(early_stopper.best_model_wts)
model.eval()
print(f'Training wall time: {(time.time()-TRAIN_START_TIME)/3600:.2f} hours')
```

```python
import torch.nn.functional as F

# Unfreeze layers for GradCAM
for param in model.backbone.parameters():
    param.requires_grad = True

class GradCAM:
    def __init__(self, model, target_layer):
        self.model = model
        self.target_layer = target_layer
        self.gradients = None
        self.activations = None
        
        target_layer.register_forward_hook(self.save_activation)
        target_layer.register_full_backward_hook(self.save_gradient)

    def save_activation(self, module, input, output):
        self.activations = output

    def save_gradient(self, module, grad_input, grad_output):
        self.gradients = grad_output[0]

    def generate_heatmap(self, input_tensor):
        model_output = self.model(input_tensor)
        self.model.zero_grad()
        model_output.backward(retain_graph=True)
        
        pooled_gradients = torch.mean(self.gradients, dim=[0, 2, 3])
        activations = self.activations.detach()[0]
        for i in range(activations.shape[0]):
            activations[i, :, :] *= pooled_gradients[i]
            
        heatmap = torch.mean(activations, dim=0).squeeze().cpu().numpy()
        heatmap = np.maximum(heatmap, 0)
        if np.max(heatmap) != 0:
            heatmap /= np.max(heatmap)
        return heatmap

# Target the deepest convolutional layer
target_layer = model.backbone.conv4 
cam = GradCAM(model, target_layer)

model.eval() 
sample_img, sample_label = next(iter(test_loader)) # Use test loader for unseen data
sample_img = sample_img[0:1].to(device)

heatmap = cam.generate_heatmap(sample_img)

# --- VISUALIZATION ---
inv_normalize = transforms.Normalize(
    mean=[-0.485/0.229, -0.456/0.224, -0.406/0.225],
    std=[1/0.229, 1/0.224, 1/0.225]
)
disp_img = inv_normalize(sample_img[0].cpu())
disp_img = disp_img.permute(1, 2, 0).numpy()
disp_img = np.clip(disp_img, 0, 1)

heatmap_resized = cv2.resize(heatmap, (299, 299))
heatmap_colored = cv2.applyColorMap(np.uint8(255 * heatmap_resized), cv2.COLORMAP_JET)
heatmap_colored = np.float32(heatmap_colored) / 255
heatmap_colored = heatmap_colored[..., ::-1] 

overlay = heatmap_colored * 0.5 + disp_img
overlay = np.clip(overlay, 0, 1)

fig, axes = plt.subplots(1, 2, figsize=(10, 5))
axes[0].imshow(disp_img)
axes[0].set_title(f"Original (True: {'Fake' if sample_label[0].item() == 1 else 'Real'})")
axes[0].axis('off')

axes[1].imshow(overlay)
axes[1].set_title("Grad-CAM Explanation")
axes[1].axis('off')

plt.show()
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
import seaborn as sns

print("Starting Final Evaluation on Test Set...")

model.eval()
all_preds = []
all_labels = []

with torch.no_grad():
    for inputs, labels in tqdm(test_loader, desc="Testing"):
        inputs = inputs.to(device)
        labels = labels.to(device).view(-1, 1).float()
        
        outputs = model(inputs)
        probs = torch.sigmoid(outputs)
        preds = (probs > 0.5).float()
        
        all_preds.extend(preds.cpu().numpy())
        all_labels.extend(labels.cpu().numpy())

acc = accuracy_score(all_labels, all_preds)
prec = precision_score(all_labels, all_preds, zero_division=0)
rec = recall_score(all_labels, all_preds, zero_division=0)
f1 = f1_score(all_labels, all_preds, zero_division=0)

print("\n--- Final XceptionNet Performance ---")
print(f"Accuracy:  {acc * 100:.2f}%")
print(f"Precision: {prec * 100:.2f}%")
print(f"Recall:    {rec * 100:.2f}%")
print(f"F1-Score:  {f1 * 100:.2f}%")

cm = confusion_matrix(all_labels, all_preds)
plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
            xticklabels=['Predicted Real', 'Predicted Fake'], 
            yticklabels=['Actually Real', 'Actually Fake'])
plt.title('XceptionNet Deepfake Detection Confusion Matrix')
plt.show()

torch.save(model.state_dict(), 'xception_deepfake_detector_final.pth')
print("Model weights successfully saved!")
```

```python
# Phase 2: ART FGSM test evaluation at BOTH 2/255 and 8/255.
# Keep this separate from the original metrics/confusion matrix above.
# Inputs to the model are ImageNet-normalized. A pixel-space perturbation eps/255
# must therefore be divided by each channel's std after normalization.
model.eval()
clean_correct = clean_total = 0
attack_epsilons_px = (2.0, 8.0)
adv_correct = {eps: 0 for eps in attack_epsilons_px}
adv_total = {eps: 0 for eps in attack_epsilons_px}
channel_std = np.asarray([0.229, 0.224, 0.225], dtype=np.float32).reshape(3, 1, 1)

# Build one ART FGSM attacker per pixel-space epsilon. Per-channel eps arrays
# account for ImageNet normalization instead of using a mean-std approximation.
ART_TEST_FGSM = {}
for eps_px in attack_epsilons_px:
    eps_normalized = (eps_px / 255.0) / channel_std
    ART_TEST_FGSM[eps_px] = FastGradientMethod(
        estimator=PyTorchClassifier(
            model=art_adapter,
            loss=nn.CrossEntropyLoss(), optimizer=optimizer,
            input_shape=(3, 299, 299), nb_classes=2,
            clip_values=(-3.0, 3.0),
            device_type='gpu' if device.type == 'cuda' else 'cpu',
        ),
        eps=eps_normalized, eps_step=eps_normalized, batch_size=16,
    )

for inputs, labels in tqdm(test_loader, desc='Clean + ART FGSM test evaluation'):
    inputs = inputs.to(device)
    labels = labels.to(device).view(-1).long()
    with torch.no_grad():
        clean_logits = model(inputs).view(-1)
        clean_preds = (torch.sigmoid(clean_logits) > 0.5).long()
    clean_correct += (clean_preds == labels).sum().item()
    clean_total += labels.numel()

    x_np = inputs.detach().cpu().numpy().astype(np.float32)
    y_np = labels.detach().cpu().numpy().astype(np.int64)
    for eps_px in attack_epsilons_px:
        adv_np = ART_TEST_FGSM[eps_px].generate(x=x_np, y=y_np)
        adv_inputs = torch.from_numpy(adv_np).to(device=device, dtype=inputs.dtype)
        with torch.no_grad():
            adv_logits = model(adv_inputs).view(-1)
            adv_preds = (torch.sigmoid(adv_logits) > 0.5).long()
        adv_correct[eps_px] += (adv_preds == labels).sum().item()
        adv_total[eps_px] += labels.numel()

print('\n--- Final Results: Clean vs ART FGSM ---')
print(f'Clean test accuracy: {100.0*clean_correct/max(1,clean_total):.2f}% ({clean_correct}/{clean_total})')
for eps_px in attack_epsilons_px:
    print(f'FGSM test accuracy ({eps_px:g}/255): {100.0*adv_correct[eps_px]/max(1,adv_total[eps_px]):.2f}% ({adv_correct[eps_px]}/{adv_total[eps_px]})')
print('Epsilon is specified in pixel space and converted per channel for ImageNet-normalized inputs.')
print('Note: both attacked accuracies use the same held-out test images and labels as clean accuracy.')
```
